# 03 - Training Dynamics & Loss Convergence\nAnalyzes L1, Perceptual, Spectral, and NDVI loss curves over training epochs.\n

```python
import json
import os
if os.path.exists('logs/training_history.json'):
    with open('logs/training_history.json') as f:
        hist = json.load(f)
    print('Logged Epochs:', len(hist))
    for h in hist[:5]:
        print(f"Epoch {h['epoch']}: Loss={h.get('train_loss', 0):.4f}, Val PSNR={h.get('val_psnr', 0):.2f}")
```
