# 01 - Sentinel-2 & WorldStrat Data Exploration\nInspects 4-band Sentinel-2 L2A BOA reflectance, band distributions, and geographic metadata.\n

```python
import numpy as np
from satellite_srm.geospatial.geotiff import read_geotiff

raster = read_geotiff('data/raw/sentinel2/S2A_MSIL2A_20260515_agriculture.tif')
print('Raster Dimensions:', raster.width, 'x', raster.height)
print('Band Count:', raster.bands, raster.metadata.band_names)
print('GSD:', raster.metadata.gsd_x, 'meters')
```

```python
# Plot RGB Composite
import matplotlib.pyplot as plt
rgb = np.stack([raster.data[2], raster.data[1], raster.data[0]], axis=-1)
rgb_norm = (rgb - rgb.min()) / (rgb.max() - rgb.min() + 1e-6)
plt.figure(figsize=(6, 6))
plt.imshow(rgb_norm)
plt.title('Sentinel-2 10m True Color (B04-B03-B02)')
plt.axis('off')
plt.show()
```
