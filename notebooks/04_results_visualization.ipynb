# 04 - Super-Resolution Results & Uncertainty Analysis\nCompares 10m input, sub-4m SR product, and pixel-level Monte Carlo uncertainty maps.\n

```python
from satellite_srm.geospatial.geotiff import read_geotiff
import matplotlib.pyplot as plt

sr_raster = read_geotiff('outputs/sr/SR_product.tif')
unc_raster = read_geotiff('outputs/uncertainty/uncertainty_map.tif')

fig, axes = plt.subplots(1, 2, figsize=(12, 6))
axes[0].imshow(sr_raster.data[2], cmap='viridis')
axes[0].set_title('Super-Resolved Sub-4m (Band 4 Red)')
axes[1].imshow(unc_raster.data[0], cmap='magma')
axes[1].set_title('Uncertainty Map (MC StdDev)')
plt.tight_layout()
plt.show()
```
